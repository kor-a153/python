## 8. 삼항연산자

```python
bool("True" if 11 == 10 else "false")
```

```python
#삼항 두수 비교
num1 = input("첫번쨰 정수\n")
num2 = input("두번쨰 정수\n")

print (num1 if num1 > num2 else "두수가 같습니다 " if num1 == num2 else num2)
```

```python
# 삼항연산자 엔터치면 안됨.
# 들여쓰기로 영역을 구분한다
num1 if num1 > num2 else "두수가 같습니다 " if num1 == num2 else num2
```
