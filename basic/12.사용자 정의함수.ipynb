## 값을 몇개나 전달할지 모를 때

```python
# 값을 몇개를 전달해도 모두 더하는 함수 : args
# 가변 매개변수, 가변인자, rest parameter

def total(*args):
    print(args)
    print(type(args))
    return sum(args)
```

```python
total(10,20)
```

```python
# 두 수를 전달하면 사칙연산의 결과 반환
def calc(num1,num2):
    return num1+ num2 , num1 - num2 , num1 * num2 , num1 / num2
```

```python
calc(10,20)
```

```python
# 언패킹
add, sub, mul, div = calc(10,20)
```

```python
print(add)
print(sub)
print(mul)
print(div)
```

## keywards argument

```python
def order_info(*args,**keywards):
    # key =value -> dict 로 묶어온게 키와즈(keywards)
    if "name" in keywards:
        print(keywards["name"])
```

```python
user_info ={
    "name": "hgd",
    "address": "강남",
    "phone": "01075698945"
}
```

```python
order_info(a="10",b="20",c="30")
```

```python
order_info()
```

```python
# 이름, 주소, 연락처, 총 구매가격
order_info(1000,50000,15000,7000,**user_info)
```

```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(**kwards):
    kor, eng, math = kwards.get("kor", 0), kwards.get("eng", 0), kwards.get("math", 0)
    total = kor + eng + math

    average = total / 3
    
    if "round" in kwards:
        average = round(average, kwards["round"])
    
    if "name" in kwards:
        name = kwards["name"]
        return f"{name}님의 평균은: {average}"

    return f"평균은: {average}"
```

```python
user_info = {
    "kor": 100,
    "eng": 80,
    "math": 22,
    "name": "홍길동",
    "round": 5
}
```

```python
get_average(**user_info)
```
