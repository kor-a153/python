## 커스텀예외 (Custom Exception)

```python
# Excetpion : 모든 예외의 부모 클래스
class BadWordException(Exception):
    def __init__(self, message):
        super().__init__(message)
```

```python
# 사용자가 "바보"라고 입력을 하면 커스텀 예외 발생!
try:
    expression = input("채팅을 입력하세요")

    if "바보" in expression:
        raise BadWordException("바보라고 하지마라")
except BadWordException as e :
    print(e)
```

```python
# 국어 , 영어 ,수학 점수를 받서 총점과 평균을 구해주는 프로그램
# 0 ~ 100 점 까지만 입력할 수 있습니다.
# 이름을 입력 할 수도 있습니다.
# 학번을 입력 할 수도 있습니다.
# 평균의 소숫점 자리를 입력받아 처리할 수 도 있습니다.
# 예외가 발생한다면 try except 문법으로 해결하세요.
```

```python
try:
    name = input("이름 입력(선택): ")
    student_id = input("학번 입력(선택): ")

    kor = int(input("국어 점수 입력(0~100): "))
    if kor < 0 or kor > 100:
        raise ValueError("국어 점수는 0~100 사이만 가능합니다.")

    eng = int(input("영어 점수 입력(0~100): "))
    if eng < 0 or eng > 100:
        raise ValueError("영어 점수는 0~100 사이만 가능합니다.")

    math = int(input("수학 점수 입력(0~100): "))
    if math < 0 or math > 100:
        raise ValueError("수학 점수는 0~100 사이만 가능합니다.")

    digit = input("평균 소수점 자리수 입력(선택): ")

    total = kor + eng + math
    avg = total / 3

    print("\n===== 성적 결과 =====")

    if name != "":
        print(f"이름 : {name}")

    if student_id != "":
        print(f"학번 : {student_id}")

    print(f"국어 : {kor}")
    print(f"영어 : {eng}")
    print(f"수학 : {math}")

    print(f"총점 : {total}")

    if digit == "":
        print(f"평균 : {avg}")
    else:
        print(f"평균 : {avg:.{int(digit)}f}")

except ValueError as e:
    print("입력 오류 :", e)

except Exception:
    print("알 수 없는 오류가 발생했습니다.")
```
