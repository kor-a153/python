```python
with open(file_path, "w") as f:
    f.write("""
        {
           "_type":"prompt",
            "template":"{country}에 대해서 어떻게 알고 있어요?",
            "input_variables":["country"]
        }
    """)
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import load_prompt
```

```python
chat = ChatOpenAI(temperature = 0.1)
```

```python
prompt = load_prompt(file_path)
```

```python
chain =prompt | chat
```

```python
result = chain.invoke({
    "country":"한국"
})
```

```python
result.content
```

## YAML

```python
file_path = os.path.join(base_path, "prompt02.yaml")
```

```python
file_path
```

```python
with open(file_path,"w") as f:
    f.write("""
"_type":"prompt"
"template":"{country}에서 가장 유명한 음식은??"
"input_variables":["country"]
    """)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = PromptTemplate.from_file(file_path)
```

```python
prompt.format(country="일본")
```

```python
chain =prompt | chat
```

```python
result = chain.invoke({
    "country":"일본"
})
```

```python
print(result.content)
```
