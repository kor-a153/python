## 컴프리헨션

```python
[ i * 2 for i in [10, 20 ,30]] # 리스트 컴프리헨션
```

```python
{ i * 2 for i in [10, 20 ,30]} # 중괄호... 뭐더라
```

```python
name = "장보고", "홍길동" , "이순신"
```

```python
name[0]
```

```python
for i in name:
    print(i)
```

```python
#enumerate : 인덱스를 만드는 함수
#[(0,장보고 ),(1 ,홍길동), (2, 이순신)] 
# unpacking
for index, name in enumerate(name):
    print (f"{index+1}번 {name}")
    
```

```python
#일반 for문(range)
# range(10) ~9
# range(10,20) -> 10~ 19
# range(20,30,2) -> 20~29, 2개씩 띄워서

"a"[0:]
"a"[10:20]
"a"[10:20:2]

for i in range(10): 
    print(i)
```

```python
# 1~50 홀수만
for i in range(1,50,2):
 print(i)
```

```python
# 1~ 50까지 홀수만 리스트에 담아서 출력
# [ i * 2 for i in [10, 20 ,30]] # 리스트 컴프리헨션

[i for i in range(1,50,2)]
# 그냥 이 자체를 리스트에 담으면 됨.
```

```python
# 1~50 까지 반복해서
# 세트 컴프리헨션에 0,1,2,3,4,5만 담아서 출력
```

```python
for i in range(1, 50):
    if i <= 5:
        print(i)
```

## 딕셔너리 빠른 for문

```python
foods = {
    "카레" , "라면", "불고기" , "떡볶이"
}

print(foods)
```

```python
for i in list(foods.values()):
    print (i)
```

```python
for key, food in foods.items():
    print(key,food)
```
