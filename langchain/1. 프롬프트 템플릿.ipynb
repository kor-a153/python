## OutParser

```python
 # LLM -> 파이썬 List
from langchain_core.output_parsers.base import BaseOutputParser

#Outparser를 제작하기 위해선 BaseOutpitParser를 상속받아야한다.
```

```python
class CommaOutputParser(BaseOutputParser):
# parse() :  반드시 재정의
def parse(self,text):
    return list(map (str.strip, text.split(",")))
```

```python
comma_parser = CommaOutputParser()
```

```python
list(map(str.strip,"a ,b ,c , d, ,e,f))
```

```python
comma_parser.parse("")
```
