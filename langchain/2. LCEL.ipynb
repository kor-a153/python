```python
class NewLineOutputParser(BaseOutputParser):
    # 반드시 parse
    def parse(self,text):
        lines = text.split("\n")
        return [line.lstirp("-123456789.").strip() for line in lines]
```

```python
newline_parser = NewLineOutputParser()
```

```python
newline_parser.parse("""-1. 햄버거\n-2. 떡볶이\n-3. 치킨""")
```

```python
template = ChatPromptTemplate.from_message([
    ("system", """
        리스트를 생성하는 기계입니다.
        요청한 모든 리스트에 개수는 최대 {max_length}개 까지만 목록으로 표시하세요.
        그 이상 초과되는 리스트는 답변하지 마세요.
    """)
    ("human", "{question}")
])

prompt = template.format_messages(
    max_length=5,
    question="AI를 잘하려면 어떤것부터 공부해야해 ?"
)
```

```python
chat.invoke
```

## 체인(chain) 생성
    - "|" 파이프연산자로 체인을 만든다.

```python
# list
first_chain = template | chat | NewLineOutputParser()
```

```python
first_chain.invoke({
    "max_length" : 5,
    "question" : "AI를 잘하려면 어떤것부터 공부해야해 ?"
})
```

```python
# 저녁메뉴 추천(양식, 중식 ,한식 , 일식 선택할수 있게)
# 체인을 생성 후 결과를 출력
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)

template = """
너는 저녁 메뉴 추천 전문가야.

사용자가 선택한 음식 종류는 {food_type}이야.

조건:
- 저녁 메뉴 3개 추천
- 각 메뉴마다 간단한 이유 작성
- 너무 길게 쓰지 말 것
- 한국어로 답변

결과 형식:
1. 메뉴명 - 추천 이유
2. 메뉴명 - 추천 이유
3. 메뉴명 - 추천 이유
"""

prompt = PromptTemplate.from_template(template)

chain = prompt | llm | StrOutputParser()

food_type = input("음식 종류를 선택하세요(양식/중식/한식/일식): ")

result = chain.invoke({
    "food_type": food_type
})

print(result)
```
